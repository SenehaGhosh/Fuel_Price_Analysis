```python
import pandas as pd
```

```python
petrol=pd.read_csv("petrol.csv")
diesel=pd.read_csv("diesel.csv")
```

```python
petrol.head()
```

```python
diesel.head()
```

```python
petrol["year"]=petrol["date"].head()
diesel["year"]=diesel["date"].head()
```

```python
petrol.info()
```

```python
petrol["date"]=pd.to_datetime(petrol["date"],format="%Y-%m-%d")
diesel["date"]==pd.to_datetime(diesel["date"],format="%Y-%m-%d")
```

```python
petrol.info()
```

```python
diesel["date"] = pd.to_datetime(diesel["date"], errors="coerce")
diesel["year"] = diesel["date"].dt.year
petrol["year"] = petrol["date"].dt.year
```

```python
diesel.info()
```

```python
diesel["date"].head(10)
```

```python
petrol["date"].head(10)
```

```python
petrol["year"]=petrol["date"].dt.year
diesel["year"]=diesel["date"].dt.year
```

```python
petrol_year = petrol.groupby("year")["rate"].mean().reset_index()
diesel_year = diesel.groupby("year")["rate"].mean().reset_index()
print(petrol_year)
print(diesel_year)
```

```python
petrol.info()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
plt.plot(petrol_year["year"], petrol_year["rate"], marker="o", label="Petrol")
plt.plot(diesel_year["year"], diesel_year["rate"], marker="s", label="Diesel")
plt.title("Year-wise Average Fuel Price in India")
plt.xlabel("Year")
plt.ylabel("Average Price (₹/Litre)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
petrol_year["Price Increase"] = petrol_year["rate"].diff()
diesel_year["Price Increase"] = diesel_year["rate"].diff()
print(petrol_year)
print(diesel_year)
```

```python
plt.figure(figsize=(10,5))
plt.bar(petrol_year["year"], petrol_year["rate"], alpha=0.7, label="Petrol")
plt.bar(diesel_year["year"], diesel_year["rate"], alpha=0.5, label="Diesel")
plt.xlabel("Year")
plt.ylabel("Average Price (₹/Litre)")
plt.title("Year-wise Average Petrol and Diesel Price")
plt.legend()
plt.show()
```

```python
state_petrol = petrol.groupby("state")["rate"].mean().sort_values(ascending=False)
```

```python
state_diesel = diesel.groupby("state")["rate"].mean().sort_values(ascending=False)
```

```python
plt.savefig("yearwise_fuel_price.png", dpi=300, bbox_inches="tight")
```

```python
yearly_petrol = petrol.groupby("year")["rate"].mean().reset_index()
yearly_diesel = diesel.groupby("year")["rate"].mean().reset_index()
```

```python
yearly_petrol.to_csv("yearly_petrol.csv", index=False)
yearly_diesel.to_csv("yearly_diesel.csv", index=False)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12,6))
state_petrol.plot(kind="bar", color="skyblue")
plt.title("State-wise Average Petrol Price")
plt.xlabel("State")
plt.ylabel("Average Price (₹/Litre)")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))
state_diesel.plot(kind="bar", color="orange")
plt.title("State-wise Average Diesel Price")
plt.xlabel("State")
plt.ylabel("Average Price (₹/Litre)")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
plt.savefig("statewise_petrol.png", dpi=300, bbox_inches="tight")
plt.savefig("statewise_diesel.png", dpi=300, bbox_inches="tight")
```

**Dataset**

Fuel price dataset containing petrol and diesel prices.

Features: City, State, Date, Fuel Price.

Python

Pandas

Matplotlib

Google Colab



---

Loaded the dataset.

Converted the date column into datetime format.

Removed unnecessary values.

Calculated yearly average fuel prices.



---

**Visualizations**

Year-wise average petrol price.

Year-wise average diesel price.

Comparison of petrol and diesel prices.

State-wise average petrol price.

State-wise average diesel price.



---

**Conclusion**

Petrol prices remained higher than diesel prices throughout the period.
